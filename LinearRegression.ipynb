```python
import numpy as np

x_train=x_train.values
y_train=y_train.values
x_train_mean = np.mean(x_train)
y_train_mean = np.mean(y_train)

num = 0
den = 0
for i in range(len(x_train)):
    num += (x_train[i] - x_train_mean)*(y_train[i] - y_train_mean)
    den += ((x_train[i] - x_train_mean)**2)
beta_1 = num/den
beta_0 = y_train_mean - (beta_1 * x_train_mean)

print(beta_1, beta_0)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('/content/drive/My Drive/Dataset/dataset.csv')
x=df.iloc[:,2]
y=df.iloc[:,3]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error, r2_score
regressor=LinearRegression()
x_train=x_train.values
y_train=y_train.values
x_train=x_train.reshape(-1,1)
y_train=y_train.reshape(-1,1)
regressor.fit(x_train,y_train)
```

 Compute the coefficient using Linear Regression function available in scikit learn library

```python
beta_0,beta_1=regressor.intercept_,regressor.coef_
print(beta_0,beta_1)
```

```python
y_pred=beta_0+beta_1*5
y_pred
```

Compute R2 value using the following equation:

```python
r_2=((y_pred-y_test.mean())**2).sum()/((y_test-y_test.mean())**2).sum()
r_2
```


Compute R2 using scikit learn function

```python
import numpy as np
x_test=x_test.values
y_test=y_test.values
x_test = x_test.reshape(-1, 1)
y_test = y_test.reshape(-1, 1)
y_predic = regressor.predict(x_test)
```
